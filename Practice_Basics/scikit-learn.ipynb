```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1. THE DATA: Real Estate
# Features: [Size in sqft, Age in years]
# Target: 1 (Expensive/Over $500k), 0 (Affordable/Under $500k)
data = {
    'sqft': [1200, 2500, 1800, 850, 3000, 1100, 2100, 1500, 4000, 2700],
    'age':  [30, 5, 15, 50, 2, 40, 10, 20, 1, 8],
    'is_expensive': [0, 1, 0, 0, 1, 0, 1, 0, 1, 1]
}

df = pd.DataFrame(data)

# 2. FEATURE SELECTION
X = df[['sqft', 'age']] #X - inputs
y = df['is_expensive'] #Y - target (result)

# 3. SPLIT DATA - 25% for testing, 75% for learning
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# 4. CHOOSE AND TRAIN THE MODEL
model = RandomForestClassifier(n_estimators=120)
model.fit(X_train, y_train) #Train the models

# 5. MAKE PREDICTIONS
# We test it on the 3 houses the model hasn't seen yet
predictions = model.predict(X_test) #Test the models

# 6. EVALUATE
acc = accuracy_score(y_test, predictions)

print(f"Accuracy: {acc:.0%}") # Turns 0.85 into 85%
print(f"Mistakes: {(y_test != predictions).sum()} out of {len(y_test)}")

# 7. USE THE MODEL ON A NEW HOUSE
# A 2200 sqft house that is 12 years old
new_house = pd.DataFrame([[2200, 12]], columns=['sqft', 'age'])
outcome = model.predict(new_house)

print(f"Prediction for new house: {'EXPENSIVE' if outcome[0] == 1 else 'AFFORDABLE'}")
```
